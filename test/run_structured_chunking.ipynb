## Separación estructural

```python
from preprocessing.json_structured_chunker import chunk_document
from preprocessing.merge_chunks import merge_structured_chunks
```

```python
with open(INPUT_DIR / "GUIDELINE - Byrne et al. - 2025 - CARCSACI Practice Guidance for Contrast Media Hypersensitivity.json", "r", encoding="utf-8") as f:
    document = json.load(f)

original_chunks = chunk_document(
    doc=document,
    doc_id="EAU",
    max_chunk_chars=2000,
)

print(
    f"Número de chunks originales: "
    f"{len(original_chunks)}"
)
```

```python
for chunk in original_chunks:
    print("=" * 100)
    print(f"Chunk ID: {chunk.chunk_id}")
    print(f"Documento: {chunk.doc_id}")
    print(f"Tipo: {chunk.chunk_type}")
    print(f"Path: {chunk.path}")
    print(f"Longitud: {len(chunk.text)} caracteres")
    print(f"Metadatos: {chunk.metadata}")
    print("-" * 100)
    print(chunk.text)
    print()
```

```python
chunk_sizes = [len(chunk.text) for chunk in original_chunks]

print("=" * 100)
print("RESUMEN DE TAMAÑOS")
print(f"Tamaño mínimo: {min(chunk_sizes)} caracteres")
print(f"Tamaño máximo: {max(chunk_sizes)} caracteres")
print(f"Tamaño medio: {sum(chunk_sizes) / len(chunk_sizes):.2f} caracteres")
```

### Postprocesamiento: unión de chunks por clave principal o tipo de chunk

```python
merged_chunks = merge_structured_chunks(
    original_chunks=original_chunks,
    doc_id="EAU",
    max_chunk_length=3000,
)

print(
    f"Número de chunks después del postprocesamiento: "
    f"{len(merged_chunks)}"
)
```

```python
for chunk in merged_chunks:
    print("=" * 100)
    print(f"Chunk ID: {chunk.chunk_id}")
    print(f"Documento: {chunk.doc_id}")
    print(f"Tipo: {chunk.chunk_type}")
    print(f"Path: {chunk.path}")
    print(f"Longitud: {len(chunk.text)} caracteres")
    print(f"Metadatos: {chunk.metadata}")
    print("-" * 100)
    print(chunk.text)
    print()
```

```python
chunk_sizes = [len(chunk.text) for chunk in merged_chunks]

print("=" * 100)
print("RESUMEN DE TAMAÑOS")
print(f"Tamaño mínimo: {min(chunk_sizes)} caracteres")
print(f"Tamaño máximo: {max(chunk_sizes)} caracteres")
print(f"Tamaño medio: {sum(chunk_sizes) / len(chunk_sizes):.2f} caracteres")
```
